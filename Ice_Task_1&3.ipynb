```python
# Check missing values in the training dataset
missing_values = train_df.isnull().sum()

# Calculate percentage of missing values
missing_percentage = (missing_values / len(train_df)) * 100

missing_values[missing_values > 0], missing_percentage[missing_percentage > 0]
```

```python
# Fill missing values
train_df.fillna({
    'workclass': 'Unknown',
    'occupation': 'Unknown',
    'native-country': 'Unknown'
}, inplace=True)

# Drop duplicates
train_df.drop_duplicates(inplace=True)

# Strip whitespace from string columns to handle inconsistencies
for col in train_df.select_dtypes(include='object').columns:
    train_df[col] = train_df[col].str.strip()

# Check for any remaining missing values and duplicates
missing_after_fill = train_df.isnull().sum()
duplicate_count = train_df.duplicated().sum()

# Check unique values to detect inconsistencies
inconsistent_summary = {col: train_df[col].unique() for col in train_df.select_dtypes(include='object').columns}

missing_after_fill, duplicate_count, inconsistent_summary
```

```python
from sklearn.preprocessing import LabelEncoder

# Create a copy for encoding
encoded_df = train_df.copy()

# Encode all categorical columns using LabelEncoder
label_encoders = {}
for col in encoded_df.select_dtypes(include='object').columns:
    le = LabelEncoder()
    encoded_df[col] = le.fit_transform(encoded_df[col])
    label_encoders[col] = le

# Normalize numeric features
from sklearn.preprocessing import StandardScaler

# Normalize all numeric features including previously clipped ones
scaler = StandardScaler()
numeric_columns = encoded_df.select_dtypes(include=['int64', 'float64']).columns.tolist()
numeric_columns.remove('income_>50K')  # exclude target

encoded_df[numeric_columns] = scaler.fit_transform(encoded_df[numeric_columns])

# Final cleaned, encoded, and normalized dataset is ready
encoded_df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
import numpy as np

# Fill missing values in train dataset
train_df.fillna({
    'workclass': 'Unknown',
    'occupation': 'Unknown',
    'native-country': 'Unknown'
}, inplace=True)

# Remove duplicate rows
train_df.drop_duplicates(inplace=True)

# Reset index after dropping duplicates
train_df.reset_index(drop=True, inplace=True)

# Handle inconsistent data: strip whitespace from categorical values
for col in train_df.select_dtypes(include='object').columns:
    train_df[col] = train_df[col].str.strip()

# Encode categorical variables using Label Encoding for correlation computation
encoded_df = train_df.copy()
label_encoders = {}
for col in encoded_df.select_dtypes(include='object').columns:
    le = LabelEncoder()
    encoded_df[col] = le.fit_transform(encoded_df[col])
    label_encoders[col] = le

# Compute correlation matrix
correlations = encoded_df.corr()['income_>50K'].abs().sort_values(ascending=False)

# Select top 4 correlated features (excluding the target)
top_features = correlations.drop('income_>50K').head(4).index.tolist()

top_features, correlations[top_features]
```

```python
# Undersample the majority class to match minority class
minority_df = train_df[train_df['income_>50K'] == 1]
majority_df = train_df[train_df['income_>50K'] == 0].sample(len(minority_df), random_state=42)

# Combine undersampled majority with minority class
balanced_df = pd.concat([minority_df, majority_df]).sample(frac=1, random_state=42).reset_index(drop=True)

# Encode categorical variables again for balanced dataset
balanced_encoded = balanced_df.copy()
for col in balanced_encoded.select_dtypes(include='object').columns:
    le = LabelEncoder()
    balanced_encoded[col] = le.fit_transform(balanced_encoded[col])

# Prepare features and target
X = balanced_encoded[top_features]
y = balanced_encoded['income_>50K']

# Normalize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split
X_train, X_val, y_train, y_val = train_test_split(X_scaled, y, test_size=0.25, random_state=42, stratify=y)

# Train logistic regression model
model = LogisticRegression()
model.fit(X_train, y_train)

# Predict and evaluate on validation set
y_pred = model.predict(X_val)
conf_matrix = confusion_matrix(y_val, y_pred)
report = classification_report(y_val, y_pred, output_dict=True)
accuracy = accuracy_score(y_val, y_pred)

conf_matrix, accuracy, report
```
